# 2.17 - Buoc 4: Doc file pcap bang Scapy
File pcap = Global Header + Packet Header (timestamp, do dai) + Packet Data.
Moi goi trich: IP src/dst, port src/dst, proto TCP/UDP/ICMP, co TCP, size, timestamp.

```python
from scapy.all import rdpcap, IP, TCP, UDP, ICMP
from pathlib import Path
BASE = Path().resolve().parent if Path().resolve().name == "notebooks" else Path().resolve()
pcap1 = BASE / "data" / "raw" / "attack-injection-hao123.pcap"
pcap2 = BASE / "data" / "raw" / "attack-injection-id1cn.pcap"
pkts = rdpcap(str(pcap1))  # Scapy khong nhan pathlib.Path, phai ep sang str
print(f"Tong so goi: {len(pkts)}")
for i, p in enumerate(pkts[:10]):
    print(i, float(p.time), p.summary(), len(p))
```
